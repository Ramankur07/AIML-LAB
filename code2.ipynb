```python
from collections import deque

def solve_water_jug_bfs(jug_a_capacity, jug_b_capacity, target_amount):
    # States are represented as (water_in_A, water_in_B)
    initial_state = (0, 0)
    # Queue for BFS, storing (state, path_to_state)
    queue = deque([(initial_state, [])])
    visited = set()
    visited.add(initial_state)

    while queue:
        current_state, path = queue.popleft()
        water_a, water_b = current_state

        # Check if target is reached
        if water_a == target_amount or water_b == target_amount:
            return path + [(current_state, "Target Reached")]

        # Possible operations
        next_states = []

        # 1. Fill Jug A
        if water_a < jug_a_capacity:
            next_states.append(((jug_a_capacity, water_b), "Fill Jug A"))
        # 2. Fill Jug B
        if water_b < jug_b_capacity:
            next_states.append(((water_a, jug_b_capacity), "Fill Jug B"))
        # 3. Empty Jug A
        if water_a > 0:
            next_states.append(((0, water_b), "Empty Jug A"))
        # 4. Empty Jug B
        if water_b > 0:
            next_states.append(((water_a, 0), "Empty Jug B"))

        # 5. Pour A to B
        # Amount to pour is min(water_a, jug_b_capacity - water_b)
        pour_amount_ab = min(water_a, jug_b_capacity - water_b)
        if pour_amount_ab > 0:
            next_states.append(((water_a - pour_amount_ab, water_b + pour_amount_ab), "Pour A to B"))

        # 6. Pour B to A
        # Amount to pour is min(water_b, jug_a_capacity - water_a)
        pour_amount_ba = min(water_b, jug_a_capacity - water_a)
        if pour_amount_ba > 0:
            next_states.append(((water_a + pour_amount_ba, water_b - pour_amount_ba), "Pour B to A"))

        for next_state, action in next_states:
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [(current_state, action)]))

    return None # No solution found

# Problem parameters
jug_a_cap = 4 # litres
jug_b_cap = 3 # litres
target = 2 # litres

print(f"Solving Water Jug Problem: Jug A ({jug_a_cap}L), Jug B ({jug_b_cap}L), Target: {target}L\n")

solution_path = solve_water_jug_bfs(jug_a_cap, jug_b_cap, target)

if solution_path:
    print("Solution Found (Minimum Steps):")
    for i, (state, action) in enumerate(solution_path):
        water_a, water_b = state
        if action == "Target Reached":
            print(f"Step {i}: Current State = ({water_a}L, {water_b}L) -> {action}")
        else:
            print(f"Step {i}: Current State = ({water_a}L, {water_b}L) -> Action: {action}")
else:
    print("No solution found for the given problem.")
```
